## 1 - Introduction to Pandas

### What is Pandas?
Pandas is a Python library that provides two core data structures:

| Structure | Dimensions | Think of it as |
|-----------|-----------|----------------|
| `Series` | 1-D | A single column with labels |
| `DataFrame` | 2-D | A table — like an Excel sheet or SQL table |

### Pandas vs NumPy vs Python Lists
| Feature | Python List | NumPy Array | Pandas DataFrame |
|---------|------------|-------------|------------------|
| Labels/index | ❌ | ❌ | ✅ (row + column names) |
| Mixed types | ✅ | ❌ | ✅ (per column) |
| Missing data | Manual | `np.nan` | Built-in `NaN` support |
| Read CSV/Excel | Manual | Limited | ✅ One line |
| GroupBy/pivot | Manual | Manual | ✅ Built-in |

## Series

```python
# Series
s = pd.Series([10, 20, 30, 40, 50])
print('Series:')
print(s)
```

```python
s = pd.Series([10, 20, 30, 40, 50],
              index=['a', 'b', 'c', 'd', 'e'])
print('Series:')
print(s)

print(f'dtype: {s.dtype}  values: {s.values}  index: {s.index.tolist()}')
```

```python
# Series arithmetic - preserves labels
print('s * 2:')
print(s * 2)

print('s + 10:')
print(s + 10)

print('s - 10:')
print(s - 10)
```

## Dataframe

```python
# DataFrame
df = pd.DataFrame({
    'name':    ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age':     [21, 22, 20, 23, 21],
    'score':   [88.5, 72.0, 91.3, 65.8, 79.2],
})

print('DataFrame:')
print(df)
```

```python
# Key attributes
print(f'shape  : {df.shape}')         # (rows, cols)
print(f'columns: {df.columns.tolist()}')
print(f'dtypes :\n{df.dtypes}')
print(f'index  : {df.index.tolist()}')
```

## Difference between Series and DataFrame

```python
# A DataFrame is literally a dict of Series

col = df['score']               # extracting one column → Series
print(type(col), '\n', col)
```

```python
# Each column is one Series sharing the same index

print('df["name"] is a Series:', isinstance(df['name'], pd.Series))
print('df itself is a DataFrame:', isinstance(df, pd.DataFrame))
```

```python
# Series vs scalar operations

scores = pd.Series([85, 92, 78, 61, 95])
print('Above 80?')
print(scores[scores > 80])
```
