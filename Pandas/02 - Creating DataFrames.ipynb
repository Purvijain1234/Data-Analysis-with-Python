## 2 - Creating DataFrames

### Ways to create a DataFrame
| Source | Method |
|--------|--------|
| Dict of lists | `pd.DataFrame({'col': [...]})` |
| List of dicts | `pd.DataFrame([{'a':1,'b':2}, ...])` |
| NumPy array | `pd.DataFrame(np_array, columns=[...])` |
| CSV file | `pd.read_csv('file.csv')` |
| Excel file | `pd.read_excel('file.xlsx')` |
| JSON | `pd.read_json('file.json')` |
| Clipboard | `pd.read_clipboard()` |

### Essential exploration methods
| Method / Attribute | What it shows |
|--------------------|---------------|
| `df.head(n)` | First n rows (default 5) |
| `df.tail(n)` | Last n rows |
| `df.sample(n)` | n random rows |
| `df.shape` | (rows, columns) |
| `df.info()` | Column names, dtypes, non-null counts, memory |
| `df.describe()` | Count, mean, std, min, quartiles, max |
| `df.dtypes` | Data type of each column |
| `df.columns` | Column names |
| `df.index` | Row index |
| `df.value_counts()` | Frequency of each unique value |


```python
import pandas as pd
import numpy as np
```

```python
# From dict of lists 

df1 = pd.DataFrame({
    'name':    ['Alice', 'Bob', 'Charlie', 'Eve', 'Sam', 'Ash', 'Din', 'Anna', 'Amy'],
    'age':     [21, 22, 20, 23, 24, 20, 23, 22, 21],
    'score':   [88.5, 72.0, 91.3, 80.5, 79.8, 89.2, 76.3, 87.5, 75.6]
})

print('From dict:')
print(df1)
```

```python
# From list of dicts
df2 = pd.DataFrame([
    {'name': 'Diana', 'age': 23, 'score': 65.8},
    {'name': 'Eve',   'age': 21, 'score': 79.2},
])

print('From list of dicts:')
print(df2)
```

```python
# From NumPy array
rng = np.random.default_rng(42)
arr = rng.integers(50, 100, (4, 3))

df3 = pd.DataFrame(arr, columns=['Maths', 'Science', 'English'])
df3.index = ['S001', 'S002', 'S003', 'S004']   # custom index

print('From NumPy:')
print(df3)
```

```python
# From CSV

df1.to_csv('students.csv', index=False)

df_csv = pd.read_csv('students.csv')
print('Read from CSV:')
print(df_csv)
```

```python
# head() -> show first 5 rows by default

print(df1.head())
```

```python
# tail() -> show last 5 rows 

print(df1.tail())
```

```python
# info() -> tells basic information 

print(df1.info())
```

```python
# describe() -> it tells descriptive stats

print(df1.describe())
```
