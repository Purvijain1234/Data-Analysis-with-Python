## 4 - Data Cleaning

Real-world data is messy. Data cleaning is typically **60-80% of any analysis project**.

### Missing value workflow
```
df.isna()           → boolean mask of all NaN positions
df.isna().sum()     → count of NaN per column
df.isna().mean()    → fraction of NaN per column

df.dropna()         → drop rows with ANY NaN
df.dropna(axis=1)   → drop columns with ANY NaN
df.dropna(thresh=3) → keep rows with AT LEAST 3 non-NaN values

df.fillna(0)        → fill all NaN with 0
df.fillna(df.mean()) → fill each column's NaN with its mean
df.fillna(method='ffill') → forward fill (deprecated in new Pandas, use ffill())
df.ffill()          → forward fill
df.bfill()          → backward fill
```

### Other cleaning operations
| Task | Method |
|------|--------|
| Find duplicates | `df.duplicated()` |
| Drop duplicates | `df.drop_duplicates()` |
| Rename columns | `df.rename(columns={'old':'new'})` |
| Drop columns | `df.drop(columns=['col1','col2'])` |
| Reset index | `df.reset_index(drop=True)` |
| Strip whitespace | `df['col'].str.strip()` |


````python
import pandas as pd
````

````python
df = pd.read_csv('students.csv')
````

````python
# Missing Value detection

print('NaN count per column:')
print(df.isna().sum())
````

````python
print('NaN percentage per column:')
print((df.isna().mean()*100).round(1).astype(str) + '%')
````

````python
print('Rows with at least one NaN:', df.isna().any(axis=1).sum())
````

````python
# dropna()

df_dropped = df.dropna()                          # drop any row with NaN
print(f'Original rows : {len(df)}')
print(f'After dropna  : {len(df_dropped)}')
````

````python
# fillna()

# Fill numeric columns with their column mean
score_cols = ['maths', 'science', 'english', 'compSci']
df_filled = df.copy()
for col in score_cols:
    df_filled[col] = df_filled[col].fillna(df_filled[col].mean().round(1))

print('After fillna with column mean:')
print('NaN remaining:', df_filled.isna().sum().sum())
print(df_filled[score_cols].head())
````

````python
# Duplicates

df_dup = pd.concat([df.head(3), df.head(3)])   # artificially create duplicates
print('Duplicate rows:', df_dup.duplicated().sum())
df_nodup = df_dup.drop_duplicates().reset_index(drop=True)

print('After drop_duplicates:', len(df_nodup))
````

````python
# Rename

df2 = df.rename(columns={'compSci': 'comp_sci', 'study_hours': 'study_hrs'})
print('Renamed columns:', df2.columns.tolist())
````

````python
# drop

df3 = df.drop(columns=['student_id', 'attendance'])
print('After drop:', df3.columns.tolist())
````
