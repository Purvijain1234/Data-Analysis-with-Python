## 10 - String Operations

Pandas provides a `.str` accessor on string columns that vectorises Python string methods.

### Key string methods
| Method | Description |
|--------|-------------|
| `.str.upper()` / `.lower()` | Change case |
| `.str.strip()` / `.str.lstrip()` | Remove whitespace |
| `.str.contains(pat)` | Boolean mask: contains substring |
| `.str.startswith(s)` / `.endswith(s)` | Prefix / suffix check |
| `.str.replace(old, new)` | Replace substring |
| `.str.split(sep)` | Split into list |
| `.str.extract(regex)` | Extract via regex capturing group |
| `.str.len()` | String length |
| `.str.count(pat)` | Count occurrences |
| `.str.get(i)` | Get item at position from list |
| `.str.cat(sep)` | Concatenate strings |

> 💡 All `.str` methods return `NaN` for `NaN` inputs - they don't crash.


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('students_enriched.csv')
```

```python
# Basic string operation
print('Upper:')
print(df['name'].str.upper().head())
```

```python
print('Name length:')
print(df['name'].str.len().head())
```

```python
# contains
print('Names containing "a" (case insensitive):')
mask = df['name'].str.contains('a', case=False, na=False)
print(df[mask]['name'].head())
```

```python
# str.startswith
print('Names starting with A or S:')
mask2 = df['name'].str.startswith(('A','S'))
print(df[mask2]['name'].values)
```

```python
# str.replace
# Shorten study category names
category_short = df['study_category'].str.replace('High','H').str.replace('Medium','M').str.replace('Low','L')

print('Shortened study categories:')
print(category_short.value_counts())
```

```python
# str.split + str.get
# Create a sample email column

df['email'] = df['name'].str.lower() + '@university.edu'
df['email_domain'] = df['email'].str.split('@').str.get(1)
print('Email and domain:'); print(df[['name','email','email_domain']].head())
```

```python
# str.extract - regex 
# Extract student number from student_id  (e.g. 'S001' → '001')

df['roll_no'] = df['student_id'].str.extract(r'S(\d+)')
print('Roll number extracted:')
print(df[['student_id','roll_no']].head())
```

```python
# str.cat - concatenate 

df['full_label'] = df['name'].str.cat(df['grade'], sep=' - Grade ')
print('Full label:')
print(df['full_label'].head())
```
