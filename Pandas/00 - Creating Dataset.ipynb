## Making of Dataset

```python
import pandas as pd
import numpy as np
```

```python
# Full sample dataset
np.random.seed(42)
n = 50

data = {
    'student_id': [f'S{i:03d}' for i in range(1, n+1)],
    'name': [
        'Aarav','Aanya','Aditi','Arjun','Bhavna','Chirag','Deepa','Dhruv','Divya','Esha',
        'Farhan','Gauri','Harsh','Isha','Jay','Karan','Kavya','Kunal','Lavanya','Manav',
        'Meera','Mihir','Naina','Nikhil','Nisha','Om','Pooja','Pranav','Priya','Rahul',
        'Riya','Rohan','Sakshi','Sanjay','Sara','Shiv','Shreya','Siddharth','Simran','Sneha',
        'Sonal','Tanvi','Tarun','Uma','Varun','Vidya','Vikram','Visha','Yash','Zara'
    ],
    'gender':       np.random.choice(['Female','Male'], n),
    'age':          np.random.randint(18, 24, n),
    'study_hours':  np.round(np.random.uniform(3, 15, n), 1),
    'maths':        np.clip(np.random.normal(65, 15, n), 0, 100).round(1),
    'science':      np.clip(np.random.normal(68, 13, n), 0, 100).round(1),
    'english':      np.clip(np.random.normal(72, 12, n), 0, 100).round(1),
    'compSci':      np.clip(np.random.normal(70, 14, n), 0, 100).round(1),
    'attendance':   np.clip(np.random.normal(80, 12, n), 40, 100).round(1),
}

# Introduce some missing values
for col in ['maths', 'science', 'english', 'compSci']:
    idx = np.random.choice(n, 4, replace=False)
    data[col] = pd.array(data[col], dtype=float)
    for i in idx: data[col][i] = np.nan

df = pd.DataFrame(data)
```

```python
df.to_csv('students.csv', index=False)
```
