## 7 - Adding & Modifying Data

### Adding new columns
```python
df['total']   = df['a'] + df['b']              # arithmetic
df['pass']    = df['score'] >= 40              # boolean
df['label']   = df['score'].apply(grade_fn)    # custom function
df = df.assign(bonus=df['score']*0.1)          # method chaining style
```

### apply() - the Swiss Army knife
```python
df['col'].apply(func)        # apply func to each element of a Series
df.apply(func, axis=0)       # apply func to each column (Series)
df.apply(func, axis=1)       # apply func to each row (Series)
```

### map() vs apply()
| | `map()` | `apply()` |
|-|---------|----------|
| Works on | Series only | Series or DataFrame |
| Input to func | Single element | Element or Series |
| Use for | Simple element-wise | Complex row/column logic |

### cut() and qcut() - binning
```python
pd.cut(df['age'], bins=[0,18,25,60,100],
       labels=['child','young','adult','senior'])  # fixed-width bins

pd.qcut(df['score'], q=4,
        labels=['Q1','Q2','Q3','Q4'])              # equal-frequency bins
```


````python
import pandas as pd
````

````python
df = pd.read_csv('students.csv')
````

````python
# Fill NaN for this section
score_cols = ['maths','science','english','compSci']
for col in score_cols:
    df[col] = df[col].fillna(df[col].mean())
````

````python
# Add calculated columns
df['total']   = df[score_cols].sum(axis=1).round(1)
df['average'] = df[score_cols].mean(axis=1).round(2)
df['passed']  = df['average'] >= 40

print('New columns added:')
print(df[['name','total','average','passed']].head())
````

````python
# apply() - grade function
def assign_grade(avg):
    if avg >= 85: return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 55: return 'C'
    elif avg >= 40: return 'D'
    return 'F'

df['grade'] = df['average'].apply(assign_grade)
print('Grade distribution:')
print(df['grade'].value_counts())
````

````python
# apply() row-wise
df['strongest'] = df[score_cols].apply(lambda row: row.idxmax(), axis=1)
print('Strongest subject per student:')
print(df[['name','strongest']])
````

````python
# map() - simple element-wise substitution
gender_map = {'Male': 'M', 'Female': 'F'}
df['gender_short'] = df['gender'].map(gender_map)

print('Gender short:') 
print(df[['gender','gender_short']])
````

````python
# assign() - method chaining
df = df.assign(
    study_category = pd.cut(df['study_hours'],
                            bins=[0, 6, 10, 15],
                            labels=['Low','Medium','High'])
)

print('Study category distribution:')
print(df['study_category'].value_counts())
````

````python
# pd.cut() - binning ages
df['age_group'] = pd.cut(df['age'],
                         bins=[17, 19, 21, 24],
                         labels=['18-19','20-21','22-23'])

print('Age groups:') 
print(df['age_group'].value_counts())
````

````python
# pd.qcut() - quartiles of average score
df['score_quartile'] = pd.qcut(df['average'], q=4, labels=['Q1','Q2','Q3','Q4'])

print('Score quartile counts:') 
print(df['score_quartile'].value_counts())
````

````python
df.to_csv('students_enriched.csv', index=False)
print('\nEnriched dataset saved.')
````
