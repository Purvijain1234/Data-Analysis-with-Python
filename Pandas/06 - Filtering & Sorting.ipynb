## 6 - Filtering & Sorting

### Boolean filtering - the core pattern
```python
df[df['col'] > value]                          # single condition
df[(df['a'] > 5) & (df['b'] == 'X')]           # AND
df[(df['a'] < 5) | (df['b'] == 'Y')]           # OR
df[~(df['col'].isna())]                        # NOT
```

### query() - readable string syntax
```python
df.query('maths > 80 and gender == "Female"')  # cleaner than chaining &
df.query('city in ["Delhi", "Mumbai"]')
df.query('50 < maths < 80')                    # chained comparison
```

### Useful filter methods
| Method | Use case |
|--------|----------|
| `isin([v1, v2])` | Match any value in list |
| `between(lo, hi)` | Inclusive range filter |
| `str.contains(pat)` | Substring filter on strings |
| `notna()` / `isna()` | Not-null / null filter |

### Sorting
```python
df.sort_values('score')                        # ascending
df.sort_values('score', ascending=False)       # descending
df.sort_values(['city', 'score'])              # multi-key sort
df.sort_index()                                # sort by index
df.nlargest(5, 'score')                        # top 5 rows
df.nsmallest(5, 'score')                       # bottom 5 rows
```


````python
import pandas as pd
````

````python
df = pd.read_csv('students.csv')
````

````python
# Boolean Filtering

print('Students with maths >= 85: ')
print(df[df['maths'] >= 85][['name', 'maths']])
````

````python
# AND condition

print('Female students, maths > 75:')
mask = (df['gender'] == 'Female') & (df['maths'] > 75)
print(df[mask][['name','gender','maths']])
````

````python
# OR condition

print('Very low (<50) or very high (>90) maths:')
mask2 = (df['maths'] < 50) | (df['maths'] > 90)
print(df[mask2][['name','maths']])
````

````python
# Query 

print('query - maths > 80 and attendance > 75:')
result = df.query('maths > 80 and attendance > 75')
print(result[['name','maths', 'attendance']])
````

````python
# isin

print('isin - age (18-20):')
print(df[df['age'].isin([18,19,20])][['name','age']])
````

````python
# between

print('between - maths 60 to 80:')
print(df[df['maths'].between(60, 80)][['name','maths']].head())
````

````python
# Sorting

print('Top 5 by maths:')
print(df.nlargest(5, 'maths')[['name','maths','age']])
````

````python
print('Sort by age (asc) then maths (desc):')
sorted_df = df.sort_values(['age','maths'], ascending=[True, False])
print(sorted_df[['name','age','maths']].head())
````
