## 7 - Figure & Axes Customisation

### Why customise?
A chart with good styling communicates faster and more clearly. The same data can look amateur or professional depending on how it's presented.

### plt.figure() - control the canvas
```python
plt.figure(figsize=(10, 5))    # width=10 inches, height=5 inches
plt.figure(dpi=150)            # resolution - 72 screen, 150 print
plt.figure(facecolor='#F5F5F5') # background colour
```

### Essential customisation functions
| Function | What it does |
|----------|--------------|
| `plt.title('text', fontsize=14, fontweight='bold')` | Chart title |
| `plt.xlabel('text')` / `plt.ylabel('text')` | Axis labels |
| `plt.xlim(lo, hi)` / `plt.ylim(lo, hi)` | Axis range |
| `plt.xticks(positions, labels, rotation=45)` | X tick marks |
| `plt.yticks(positions, labels)` | Y tick marks |
| `plt.legend(loc='upper left')` | Legend |
| `plt.grid(True, alpha=0.3)` | Grid lines |
| `plt.tight_layout()` | Auto-fix overlapping elements |

### Removing spines (cleaner look)
```python
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
```

### Colour formats Matplotlib accepts
```python
'red'           # named colour
'#4C72B0'       # hex code
(0.3, 0.5, 0.8) # RGB tuple (values 0-1)
'C0', 'C1'      # default colour cycle
```


````python
import matplotlib.pyplot as plt
````

````python
import numpy as np

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
score_a = [60, 62, 65, 64, 68, 72, 70, 74, 76, 79, 82, 85]
score_b = [55, 57, 60, 63, 65, 67, 64, 69, 71, 74, 77, 76]
````

````python
# Plot 1 : Default (no customisation) 
plt.figure(figsize=(8, 4))
plt.plot(months, score_a, label='Group A')
plt.plot(months, score_b, label='Group B')

plt.title('Default Matplotlib Styling')
plt.legend()

plt.tight_layout()
plt.show()
````

````python
# Plot 2 : Fully customised
plt.figure(figsize=(10, 5), facecolor='#FAFAFA')

plt.plot(months, score_a, color='#2E86AB', lw=2.5, marker='o',
         ms=7, markerfacecolor='white', markeredgewidth=2, label='Group A')
plt.plot(months, score_b, color='#E84855', lw=2.5, marker='s',
         ms=7, markerfacecolor='white', markeredgewidth=2, label='Group B')

plt.fill_between(months, score_a, score_b, alpha=0.06, color='purple')

plt.title('Monthly Score Comparison', fontsize=14, fontweight='bold', pad=14, color='#222222')
plt.xlabel('Month', fontsize=11, labelpad=8)
plt.ylabel('Average Score', fontsize=11, labelpad=8)
plt.ylim(45, 95)
plt.xticks(rotation=45, fontsize=9)
plt.yticks(fontsize=9)
plt.legend(loc='upper left', fontsize=10, framealpha=0.7, fancybox=True)
plt.grid(axis='y', alpha=0.3, linestyle='--', color='gray')
plt.gca().set_axisbelow(True)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
````

````python
print('Same data - completely different feel.')
print('Professional charts: remove top/right spines, subtle grid, tight layout.')
````
