## 3 - Bar Charts

### What is a Bar Chart?
A **bar chart** uses rectangular bars to compare values across **discrete categories**. The height (or length) of each bar represents the value.

### When to use a Bar Chart
✅ Comparing categories (subjects, cities, groups)
✅ Showing ranking
✅ Comparing multiple series side-by-side (grouped)
✅ Part-to-whole where order matters (stacked)
❌ Do NOT use for continuous data over time (use line chart instead)

### Types of bar charts
| Type | Function | Best for |
|------|----------|----------|
| Vertical bar | `plt.bar()` | Default — compare categories |
| Horizontal bar | `plt.barh()` | Long category names, rankings |
| Grouped bar | Multiple `plt.bar()` with offset `x` | Compare groups side by side |
| Stacked bar | `plt.bar()` with `bottom=` | Show parts AND totals |

### Key `plt.bar()` parameters
| Parameter | What it does |
|-----------|-------------|
| `x` | Category positions (numbers or strings) |
| `height` | Bar values |
| `width` | Bar width (default 0.8) |
| `color` | Fill colour |
| `edgecolor` | Border colour |
| `alpha` | Transparency |
| `label` | Legend label |
| `bottom` | Start height (for stacked bars) |


```python
import matplotlib.pyplot as plt
```

```python
subjects    = ['Maths', 'Science', 'English', 'CompSci']
female_avg  = [66.2, 70.1, 74.5, 69.8]
male_avg    = [64.8, 67.3, 70.2, 71.5]
cities      = ['Delhi', 'Mumbai', 'Pune', 'Bangalore', 'Hyderabad']
city_avg    = [68.4, 71.2, 65.8, 73.1, 69.5]
color1      = ['#4C72B0', '#DD8452', '#55A868', '#C44E52']
color2      = ['#4C72B0','#DD8452','#55A868','#C44E52','#8172B3']
```

### Plot 1 : Vertical bar with value labels

```python
plt.bar(subjects, female_avg, color=color1, edgecolor='white', width=0.55)

plt.title('Average Score by Subject (Female Students)')
plt.ylabel('Average Score')

plt.ylim(0, 90)
plt.grid(axis='y', alpha=0.3)
        
plt.tight_layout()
plt.show()
```

### Plot 2 : Horizontal bar

```python
plt.barh(cities, city_avg, color=color2, edgecolor='white')

plt.title('Average Score by City')
plt.xlabel('Average Score')

plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Plot 3 : Grouped bar

```python
import numpy as np
x = np.arange(len(subjects))
w = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x - w/2, female_avg, width=w, color='steelblue', edgecolor='white', label='Female')
plt.bar(x + w/2, male_avg,   width=w, color='coral', edgecolor='white', label='Male')

plt.xticks(x, subjects)
plt.title('Subject Scores - Female vs Male (Grouped)')
plt.ylabel('Average Score')

plt.ylim(0, 90)
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### Plot 4 : Stacked bar

```python
plt.bar(subjects, female_avg, color='yellow', edgecolor='white', label='Female')
plt.bar(subjects, male_avg,   color='pink', edgecolor='white', label='Male', bottom=female_avg)

plt.title('Subject Scores - Stacked (Female + Male)')
plt.ylabel('Combined Score')

plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
