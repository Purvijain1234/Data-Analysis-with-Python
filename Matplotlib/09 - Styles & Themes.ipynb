## 9 - Styles & Themes

### What are Matplotlib styles?
Matplotlib comes with **built-in style sheets** that change the look of all your charts instantly — colours, fonts, grid style, background - without changing any data or code.

### How to apply a style
```python
plt.style.use('ggplot')          # apply to all charts from this point

# OR temporarily (just for one block):
with plt.style.context('seaborn-v0_8-whitegrid'):
    plt.plot(x, y)
    plt.show()
# after the 'with' block, style resets
```

### See all available styles
```python
print(plt.style.available)   # prints all style names
```

### Popular built-in styles
| Style name | Look & feel |
|------------|------------|
| `'default'` | Classic Matplotlib blue |
| `'ggplot'` | R's ggplot2 — pink background, clean lines |
| `'seaborn-v0_8-whitegrid'` | White background, subtle grid — very clean |
| `'seaborn-v0_8-darkgrid'` | Dark grid lines |
| `'fivethirtyeight'` | FiveThirtyEight news website style |
| `'bmh'` | Bayesian Methods for Hackers — muted colours |
| `'dark_background'` | Dark mode |
| `'Solarize_Light2'` | Warm yellow tones |

### plt.rcParams - global default settings
```python
plt.rcParams['figure.figsize']   = (10, 5)    # default figure size
plt.rcParams['font.size']        = 12         # default font size
plt.rcParams['axes.titlesize']   = 14         # title size
plt.rcParams['lines.linewidth']  = 2.0        # line thickness
plt.rcParams['axes.grid']        = True       # always show grid

plt.rcdefaults()   # reset everything back to defaults
```


````python
import matplotlib.pyplot as plt
````

````python
import numpy as np

rng = np.random.default_rng(7)
x   = np.linspace(0, 10, 50)
y1  = np.sin(x) + rng.normal(0, 0.1, 50)
y2  = np.cos(x) + rng.normal(0, 0.1, 50)
````

````python
# Show all styles
print('Available styles (first 15):')  
print(plt.style.available[:15])
print(f'Total styles: {len(plt.style.available)}')
````

````python
# Compare 6 popular styles
popular = ['default', 'ggplot', 'seaborn-v0_8-whitegrid',
           'fivethirtyeight', 'bmh', 'dark_background']

plt.figure(figsize=(16, 8))

for i, style in enumerate(popular):
    plt.subplot(2, 3, i + 1)
    with plt.style.context(style):
        plt.plot(x, y1, lw=2, label='sin(x)')
        plt.plot(x, y2, lw=2, label='cos(x)')
        plt.title(f"style: '{style}'", fontsize=9, fontweight='bold')
        plt.legend(fontsize=8)

plt.suptitle('Built-in Styles Comparison', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Using seaborn whitegrid with custom rcParams
with plt.style.context('seaborn-v0_8-whitegrid'):
    plt.rcParams.update({
        'axes.titlesize' : 14,
        'axes.labelsize' : 12,
        'lines.linewidth': 2.5,
        'lines.markersize': 8,
        'legend.fontsize' : 10,
    })
    plt.figure(figsize=(9, 4))
    plt.plot(x, y1, marker='o', markevery=7, label='sin(x)')
    plt.plot(x, y2, marker='s', markevery=7, label='cos(x)')
    plt.title('seaborn-whitegrid + custom rcParams')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.tight_layout()
    plt.show()

plt.rcdefaults()   # reset to Matplotlib defaults
print('rcParams reset to defaults.')
````
