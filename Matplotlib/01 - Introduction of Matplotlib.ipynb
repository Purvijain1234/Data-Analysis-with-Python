# 1 - Introduction of Matplotlib

## What is Data Visualisation?

**Data Visualisation** is the process of representing data and information through visual elements like charts, graphs, and maps.

Before visualisation, data is just numbers:
```
65, 72, 88, 45, 91, 78, 55, 83, 70, 62 ...
```

After visualisation, it becomes a **story**:
- Most students score between 60–80
- A small group scores very high (>90)
- A few students need extra support (<50)

### Why is Data Visualisation important?

| Without Charts | With Charts |
|----------------|-------------|
| Numbers are hard to compare | Patterns become instantly visible |
| Trends are hidden in rows | Trends jump out visually |
| Hard to communicate to others | Anyone can understand a good chart |
| Takes time to read a table | A chart delivers insight in seconds |

### Data Visualisation in the Data Science Pipeline
```
Collect Data  →  Clean Data  →  Explore (EDA)  →  Analyse  →  Communicate
                                     ↑                               ↑
                           Charts help you UNDERSTAND        Charts help you EXPLAIN
                           the data yourself                 findings to others
```

## What is Matplotlib?

**Matplotlib** is Python's foundational plotting library, created by John D. Hunter in 2003. It is the base on which Seaborn, Pandas plotting, and many other tools are built.

### Why Matplotlib?

| Feature | Detail |
|---------|--------|
| Free & open source | Completely free to use |
| Highly customisable | Control every pixel of a chart |
| Works everywhere | Jupyter, scripts, web apps, PDFs |
| Huge community | Tons of examples and Stack Overflow answers |
| Foundation | NumPy, Pandas, Seaborn all use it underneath |

### What can you make with Matplotlib?
```
📈 Line charts        📊 Bar charts       🔵 Scatter plots
🔔 Histograms         🥧 Pie / Donut      🌡️ Heatmaps
📦 Box plots          🎻 Violin plots     🗺️ Contour maps
📐 Subplots grids     🖼️ Image display    ⬆️ Dual Y-axis charts
```

### pyplot - the simple interface

`matplotlib.pyplot` is a collection of functions that feel like drawing commands. You call them one after another to build up your chart:

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [10, 20, 30])  # draw a line
plt.title('My First Chart')         # add a title
plt.xlabel('X Axis')                # label x axis
plt.ylabel('Y Axis')                # label y axis
plt.show()                          # display the chart
```

### The pyplot workflow - always the same pattern
```
1. plt.figure()          → create a canvas (optional but good practice)
2. plt.plot() / bar() etc → draw the chart
3. plt.title()           → add title
4. plt.xlabel / ylabel() → label the axes
5. plt.legend()          → add legend (if multiple series)
6. plt.tight_layout()    → auto-fix spacing
7. plt.savefig()         → save to file (before show!)
8. plt.show()            → display
```

> 💡 **Convention:** Always import as `plt`. Install with `pip install matplotlib`.

````python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 35, 28]

plt.figure(figsize=(8,4))  # canvas
plt.plot(x, y, color='steelblue', linewidth=2.5, marker='o', markersize=8, label='Monthly Score')  # draw
plt.title('First Chart', fontsize=14, fontweight='bold')  #title of chart
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('first_chart', dpi=120, bbox_inches='tight')  # save the chart
plt.show() # display 

print('pyplot workflow used:')
print('  plt.figure()  → canvas')
print('  plt.plot()    → draw')
print('  plt.title()   → title')
print('  plt.show()    → display')
````
