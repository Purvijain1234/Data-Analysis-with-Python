## 6 - Regression Plots (Linear & Polynomial)

### Overview
Regression plots display **relationship trends with fitted regression lines**.

### Available Functions
| Function | Purpose |
|----------|----------|
| `sns.regplot()` | Axes-level regression plot |
| `sns.lmplot()` | Figure-level with multi-panels |
| `sns.residplot()` | Residual analysis |

## Joint Plots
Bivariate plot + marginal distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
```

```python
# Linear regression
g1 = sns.lmplot(data=tips, x='total_bill', y='tip', height=5, aspect=1.3)
```

```python
# Regression by gender
g2 = sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex', height=5, aspect=1.3)
```

```python
# Joint plot
g3 = sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', height=6)
```

```python
# Joint with regression
g4 = sns.jointplot(data=tips, x='total_bill', y='tip', kind='reg', height=6)
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.regplot(data=tips, x='total_bill', y='tip', scatter_kws={'s': 50, 'alpha': 0.6}, color='steelblue')
plt.title('Linear Regression', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.subplot(1, 2, 2)
sns.regplot(data=tips, x='total_bill', y='tip', order=2, scatter_kws={'s': 50, 'alpha': 0.6}, color='coral')
plt.title('Polynomial Regression (order=2)', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.suptitle('Regression Plots', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
