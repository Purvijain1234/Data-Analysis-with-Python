## 4 - Categorical Plots (Bar, Box, Violin)

### Overview
Categorical plots display **distributions and comparisons across categorical groups**.

### Available Functions
| Function | Purpose |
|----------|----------|
| `sns.barplot()` | Average values with confidence intervals |
| `sns.countplot()` | Frequency counts |
| `sns.boxplot()` | Distribution quartiles & outliers |
| `sns.violinplot()` | Detailed distribution shape |
| `sns.stripplot()` | Individual data points (scatter) |
| `sns.swarmplot()` | Non-overlapping points |
| `sns.pointplot()` | Mean values with connections |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
```

```python
# Bar plot
g1 = sns.catplot(data=tips, x='day', y='tip', kind='bar', height=5, aspect=1.3)
```

```python
# Box plot
g2 = sns.catplot(data=tips, x='day', y='total_bill', kind='box', height=5, aspect=1.3)
```

```python
# Violin plot with hue
g3 = sns.catplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    kind='violin',
    height=5,
    aspect=1.3
)
```

```python
plt.figure(figsize=(15, 9))

plt.subplot(2, 3, 1)
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', palette='Set2')
plt.title('Bar Plot: Avg Bill by Day', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Average Bill ($)')

plt.subplot(2, 3, 2)
sns.countplot(data=tips, x='day', hue='sex', palette='husl')
plt.title('Count Plot: Frequency', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Count')

plt.subplot(2, 3, 3)
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex', palette='Pastel1')
plt.title('Box Plot: Distribution', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')

plt.subplot(2, 3, 4)
sns.violinplot(data=tips, x='day', y='total_bill', palette='Set3')
plt.title('Violin Plot: Distribution Shape', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')

plt.subplot(2, 3, 5)
sns.stripplot(data=tips, x='day', y='total_bill', hue='sex', palette='bright', jitter=True, size=7)
plt.title('Strip Plot: Individual Points', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')

plt.subplot(2, 3, 6)
sns.pointplot(data=tips, x='day', y='total_bill', hue='sex', palette='dark')
plt.title('Point Plot: Mean with CI', fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Average Bill ($)')

plt.suptitle('Categorical Plots', fontsize=24, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()
```
