## 7 - Multi-plot Grids (FacetGrid, PairGrid, JointGrid)

### Overview
Multi-plot grids create **multi-panel visualizations** for comparing across multiple variables or subsets.

### Available Functions
| Function | Purpose |
|----------|----------|
| `FacetGrid` | Wrap plots across categorical levels |
| `PairGrid` | Scatter matrix of all variable pairs |
| `JointGrid` | Bivariate + marginal distributions |
| `clustermap()` | Hierarchical clustering + heatmap |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
iris = sns.load_dataset('iris')
```

```python
# Simple pairplot
g1 = sns.pairplot(data=iris, height=2, aspect=1)
```

```python
# PairPlot with hue
g2 = sns.pairplot(data=iris, hue='species', palette='Set1', height=2, aspect=1)
```

```python
# PairPlot with different diagonal
g3 = sns.pairplot(data=iris, hue='species', diag_kind='kde', palette='husl', height=2)
```

```python
print('Creating PairGrid (may take a moment)...')
iris_subset = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']]
g = sns.PairGrid(iris_subset, hue='species', height=2.2, palette='Set1')
g.map_diag(sns.histplot, kde=True)
g.map_upper(sns.scatterplot, s=50, alpha=0.6)
g.map_lower(sns.kdeplot, fill=True)
g.fig.suptitle('PairGrid: Iris Pair Relationship', fontsize=12, fontweight='bold', y=0.995)
plt.show()
```
