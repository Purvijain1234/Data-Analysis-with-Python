## 1 - Introduction to Seaborn

### What is Seaborn?
Seaborn is a **high-level, data-aware** plotting library built on Matplotlib. It provides a simple interface for creating informative and beautiful statistical graphics.

### Why Seaborn?
- **Beautiful by default** - Elegant themes out of the box
-  **Pandas integration** - Works directly with DataFrames
-  **Statistical visualization** - Estimates and confidence intervals built-in
-  **Multi-plot grids** - Complex layouts made simple
-  **Less code** - Create sophisticated plots with minimal code

### Two API Styles
- **Axes-level** (e.g., `scatterplot()`) - Work with matplotlib axes (more control)
- **Figure-level** (e.g., `relplot()`) - Create entire figures with subplots (simpler)
