## 3 - Relational Plots (Scatter & Line)

### Overview
Relational plots show **relationships between two or more continuous variables**.

### Key Functions
- `sns.scatterplot()` - Scatter plot
- `sns.lineplot()` - Line plot
- `sns.relplot()` - Figure-level (creates new figure with subplots)

### Key Parameters
| Parameter | Purpose |
|-----------|----------|
| `x, y` | Variables to plot |
| `hue` | Color by categorical variable |
| `size` | Point size by continuous variable |
| `style` | Marker style by categorical variable |
| `data` | DataFrame with the data |
| `palette` | Color palette |
| `col, row` | Create subplots by categories |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
```

### Scatter Plot

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(data=tips, x='total_bill', y='tip', s=80, alpha=0.6, color='steelblue')
plt.title('Basic Scatter: Bill vs Tip', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.subplot(1, 2, 2)
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex',
                palette='Set1', sizes=(50, 250), alpha=0.6)
plt.title('Scatter with Hue & Size', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

### Line Plots

```python
plt.figure(figsize=(12, 5))
flights_year = flights[flights['year'] == 1949]
sns.lineplot(data=flights_year, x='month', y='passengers', marker='o', linewidth=2.5, color='steelblue')
plt.title('Monthly Airline Passengers (1949)', fontweight='bold', fontsize=13)
plt.xlabel('Month')
plt.ylabel('Number of Passengers')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Simple scatter plot
g1 = sns.relplot(data=tips, x='total_bill', y='tip', height=5, aspect=1.3)
g1.set(title='Bill vs Tip')
```

```python
# Scatter with multiple dimensions
g2 = sns.relplot(
    data=tips, 
    x='total_bill', 
    y='tip',
    hue='sex',           # Color by gender
    col='day',           # Separate by day
    height=4,
    aspect=1
)
```
