## 5 - Distribution Plots (Histogram, KDE, ECDF)

### Overview
Distribution plots reveal **how data is spread across values**.

### Available Functions
| Function | Purpose |
|----------|----------|
| `sns.histplot()` | Histogram with flexible binning |
| `sns.kdeplot()` | Kernel Density Estimation (smooth) |
| `sns.ecdfplot()` | Empirical Cumulative Distribution |
| `sns.rugplot()` | Individual data point ticks |
| `sns.displot()` | Figure-level distribution |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
```

```python
# Histogram with KDE
g1 = sns.displot(data=tips, x='total_bill', kde=True, bins=20, height=5, aspect=1.3)
```

```python
# KDE by group
g2 = sns.displot(data=tips, x='total_bill', hue='sex', kind='kde', height=5, aspect=1.3)
```

```python
# ECDF plot
g3 = sns.displot(data=tips, x='total_bill', kind='ecdf', hue='sex', height=5, aspect=1.3)
```

```python
sns.kdeplot(data=tips, x='total_bill', y='tip', fill=True, cmap='Blues')
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
sns.histplot(data=tips, x='total_bill', kde=True, bins=20, color='steelblue')
plt.title('Histogram with KDE', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')

plt.subplot(2, 2, 2)
sns.kdeplot(data=tips, x='total_bill', y='tip', fill=True, cmap='Blues')
plt.title('2D KDE Plot', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.subplot(2, 2, 3)
sns.ecdfplot(data=tips, x='total_bill', hue='sex', palette='Set1')
plt.title('ECDF: Cumulative Distribution', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Cumulative Probability')

plt.subplot(2, 2, 4)
sns.histplot(data=tips, x='total_bill', hue='sex', kde=True, bins=15, palette='husl')
plt.title('Histogram by Group', fontweight='bold')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')

plt.suptitle('Distribution Plots', fontsize=24, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()
```
