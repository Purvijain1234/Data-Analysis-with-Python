## 11 - Capstone Project: Exploratory Data Analysis

### Objective
Create a comprehensive exploratory data analysis (EDA) dashboard for the Iris dataset.

### Dashboard Components
1. **Dataset Overview** - Basic statistics and info
2. **Univariate Analysis** - Feature distributions
3. **Bivariate Analysis** - Feature relationships
4. **Multivariate Analysis** - PairPlot
5. **Correlation Analysis** - Heatmap and insights

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris_data = sns.load_dataset('iris')

print(f'\nDataset shape: {iris_data.shape}')
print(f'Column names: {list(iris_data.columns)}')
print(f'\nData types:\n{iris_data.dtypes}')
print(f'\nBasic Statistics:\n{iris_data.describe()}')
print(f'\nMissing values:\n{iris_data.isnull().sum()}')
```

```python
# Page 1: Univariate Analysis
plt.figure(figsize=(15, 4))

for i, col in enumerate(iris_data.select_dtypes(include='float64').columns, 1):
    plt.subplot(1, 4, i)
    sns.histplot(data=iris_data, x=col, kde=True, bins=20, color='steelblue')
    plt.title(f'{col} Distribution', fontweight='bold')

plt.suptitle('Univariate Analysis', fontsize=13, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()
```

```python
# Page 2: Bivariate Analysis 

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(data=iris_data, x='sepal_length', y='sepal_width', hue='species',
                size='petal_length', palette='Set1', sizes=(50, 250), alpha=0.6)
plt.title('Sepal Dimensions by Species', fontweight='bold')

plt.subplot(1, 2, 2)
sns.scatterplot(data=iris_data, x='petal_length', y='petal_width', hue='species',
                palette='Set2', s=100, alpha=0.6)
plt.title('Petal Dimensions by Species', fontweight='bold')

plt.suptitle('Bivariate Analysis', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Page 3: Multivariate Analysis - PairPlot 

g = sns.pairplot(iris_data, hue='species', palette='Set2', diag_kind='kde')
g.fig.suptitle('PairPlot: All Feature Relationships', fontsize=13, fontweight='bold', y=0.995)
plt.show()
```

```python
# Page 4: Correlation Analysis ──────────────────────────────
print('\n' + '='*70)
print('PAGE 4: CORRELATION ANALYSIS')
print('='*70)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
numeric_iris = iris_data.select_dtypes(include='float64')
corr = numeric_iris.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, square=True,
            linewidths=2, fmt='.2f', cbar_kws={'label': 'Correlation'})
plt.title('Correlation Matrix', fontweight='bold')

plt.subplot(1, 2, 2)
sns.boxplot(data=iris_data, x='species', y='sepal_length', palette='Set2')
plt.title('Sepal Length by Species', fontweight='bold')
plt.ylabel('Sepal Length (cm)')

plt.suptitle('Correlation Analysis | Section 11', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python








# ── Summary insights ──────────────────────────────────────────
print('\n' + '='*70)
print('KEY INSIGHTS FROM EDA')
print('='*70)

for species in iris_data['species'].unique():
    species_data = iris_data[iris_data['species'] == species]
    print(f'\n{species.upper()}:')
    print(f'  Count: {len(species_data)}')
    print(f'  Sepal Length: {species_data["sepal_length"].mean():.2f} ± {species_data["sepal_length"].std():.2f} cm')
    print(f'  Petal Length: {species_data["petal_length"].mean():.2f} ± {species_data["petal_length"].std():.2f} cm')

print('\n' + '='*70)
print('✓ CAPSTONE PROJECT COMPLETED')
print('='*70)
```
