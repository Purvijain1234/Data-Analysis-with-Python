## 9 - Advanced Styling & Customization

### Techniques
1. **Figure size & layout** - `plt.figure()`, `tight_layout()`
2. **Titles & labels** - Clear, bold, meaningful text
3. **Legends** - Placement, size, formatting
4. **Colors** - Custom palettes and edge colors
5. **Grids & backgrounds** - Alpha, linestyle, facecolor
6. **Annotations** - Text, arrows, highlights

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
```

```python
plt.figure(figsize=(12, 6))

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex',
               style='time', palette='Set1', sizes=(80, 300), alpha=0.7,
               edgecolor='black', linewidth=1.5)

plt.title('Advanced Customization Example', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Total Bill ($)', fontsize=12, fontweight='bold')
plt.ylabel('Tip ($)', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3, linestyle='--')

# Customize legend
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10, framealpha=0.95, edgecolor='black')

plt.tight_layout()
plt.show()
```
