## 2 - Themes, Styles & Color Palettes

### Set Style
Use `sns.set_style()` to change the overall look:
- `'darkgrid'` (default) - Dark background with grid
- `'whitegrid'` - White background with grid
- `'dark'` - Dark background, no grid
- `'white'` - White background, no grid
- `'ticks'` - Minimal with ticks only

### Set Context
Use `sns.set_context()` to scale plot elements:
- `'paper'` - Smallest (for publication)
- `'notebook'` - Default (for Jupyter)
- `'talk'` - Larger (for presentations)
- `'poster'` - Largest (for posters)

### Color Palettes
Use `sns.color_palette()` or `sns.set_palette()`:
- **Categorical**: `'deep'`, `'muted'`, `'pastel'`, `'bright'`, `'dark'`, `'colorblind'`
- **Sequential**: `'Blues'`, `'Reds'`, `'Greens'`, `'YlOrRd'`
- **Diverging**: `'coolwarm'`, `'RdBu'`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

### Styles

```python
plt.figure(figsize=(14, 10))

styles = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
x = np.linspace(0, 10, 50)

for i, style in enumerate(styles, 1):
    if i == 5:
        # Add an extra empty subplot for 5 plots
        continue
    plt.subplot(2, 3, i)
    sns.set_style(style)
    y = np.sin(x) + np.random.normal(0, 0.1, 50)
    plt.plot(x, y, marker='o', color='steelblue', markersize=4, linewidth=2)
    plt.title(f"Style: '{style}'", fontweight='bold')
    plt.xlabel('X')
    plt.ylabel('Y')

# Plot 5: ticks style
plt.subplot(2, 3, 5)
sns.set_style('ticks')
y = np.sin(x) + np.random.normal(0, 0.1, 50)
plt.plot(x, y, marker='o', color='steelblue', markersize=4, linewidth=2)
plt.title("Style: 'ticks'", fontweight='bold')
plt.xlabel('X')
plt.ylabel('Y')

plt.suptitle('Seaborn Styles', fontsize=24, fontweight='bold', y=0.995)
plt.tight_layout()
sns.set_style('whitegrid')  # Reset
plt.show()
```

### Color

```python
plt.figure(figsize=(15, 6))

palettes = ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind']

for i, pal_name in enumerate(palettes, 1):
    plt.subplot(2, 3, i)
    palette = sns.color_palette(pal_name, 6)
    colors_y = [1, 1, 1, 1, 1, 1]
    plt.bar(range(6), colors_y, color=palette)
    plt.title(f"Palette: '{pal_name}'", fontweight='bold')
    plt.ylim(0, 1.3)
    plt.xticks([])
    plt.ylabel('Color Intensity')

plt.suptitle('Seaborn Color Palettes', fontsize=24, fontweight='bold')
plt.tight_layout()
plt.show()
```
