## 3. Numpy Basics

### Key array attributes
| Attribute | Meaning |
|-----------|--------|
| `.ndim` | Number of dimensions |
| `.shape` | Sizes per axis |
| `.size` | Total elements |
| `.dtype` | Element data type |
| `.nbytes` | Total bytes |

### NumPy dtypes
| dtype | Bytes |
|-------|-------|
| `int8/16/32/64` | 1/2/4/8 |
| `float32/64` | 4/8 |
| `complex64/128` | 8/16 |
| `bool` | 1 |

```python
import numpy as np
```

```python
# Shape
arr_2d = np.array([[1,2,3],
                  [4,5,6]])

print(arr_2d.shape)
```

```python
# Size
arr_2d = np.array([[10,20,30],
                  [40,50,60]])

print(arr_2d.size)
```

```python
# Dimesnion
arr_1d = np.array([1,2,3])
arr_2d = np.array([[1,2,3], [4,5,6]])
arr_3d = np.array([[[1,2], [3,4], [5,6], [7,8]]])

print(arr_1d.ndim)
print(arr_2d.ndim)
print(arr_3d.ndim)
```

```python
# Data type & Nbytes
arr = np.array([1,2,3])
print(arr.dtype)
print(arr.nbytes)

arr = np.array([1,2.5,3])
print(arr.dtype)
print(arr.nbytes)
```

```python
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([[1,2,3],[4,5,6]])
a3 = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])

for name, arr in [('1-D',a1),('2-D',a2),('3-D',a3)]:
    print(f'{name}: ndim={arr.ndim}, shape={arr.shape}, size={arr.size}, dtype={arr.dtype}, nbytes={arr.nbytes}')
```

```python
# Astype
arr = np.array([1.2, 2.5, 3.6])
print(arr.dtype)

int_arr = arr.astype(int)

print(int_arr)
print(int_arr.dtype)
```

```python
# Type casting
x = np.array([1.7, 2.9, 3.1])
print('float64:', x)
print('as int :', x.astype(int))   # truncates
```

```python
# dtypes
print('int8  :', np.array([1,2,3],dtype=np.int8),    ' bytes=', np.array([1,2,3],dtype=np.int8).nbytes)
print('float32:', np.array([1,2,3],dtype=np.float32),' bytes=', np.array([1,2,3],dtype=np.float32).nbytes)
print('bool  :', np.array([1,0,1,0],dtype=bool))
```

```python
# maths
arr = np.array([10,20,30])

print(arr + 5)
print(arr * 2)
print(arr ** 2)
```

```python
# Aggregating Function
arr = np.array([10, 20, 30, 40, 50])

print(np.sum(arr))
print(np.mean(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
```
