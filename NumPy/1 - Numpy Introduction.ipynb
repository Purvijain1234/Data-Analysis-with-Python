## 1. Introduction of NumPy 

### NumPy provides:
- A powerful **N-dimensional array object** (`ndarray`)
- **Vectorised** mathematical operations (element-wise, no explicit loops)
- Tools for **linear algebra**, **random numbers**, and **Fourier transforms**
- A **C backend** - operations run at near-C speed

### Feature:  | Use Case:
- Faster Speed | - Data Science
- Less Memory  | - Image Processing
- Easy Math Operations | - Stock Market & Finance
- Used in AI, DA, DS, ML | - Medical Research



### Python List vs NumPy Array

| Factors | List | NumPy |
|---------|------------|------------|
| Element types | Mixed | Homogeneous |
| Speed | Slow | Fast |
| Memory | More | Less |
| Math | Manual loop | Element-wise by default (Easy)|
| Dimensions | 1-D only | N-D |
| Used In | Small data | Large data |

> 💡 Convention: always import NumPy as `np`.


```python
# for basic or small data it will be easy for loop 
# but for large data loop will slow therefore we use or apply numpy

# using loop
temperatures = [32.5, 35.8, 32.3, 36.5, 28.6]
total = 0
for temp in temperatures:
    total += temp

avg = total/len(temperatures)
print('Average Temperature: ', avg)

# Using Numpy
import numpy as np
temperatures = np.array([32.5, 35.8, 32.3, 36.5, 28.6])
avg = np.mean(temperatures)

print('Average Temperature: ', avg)
```

```python
import numpy as np
import time, sys

# Speed comparison: list vs array
SIZE = 1_000_000
py_list = list(range(SIZE))
np_arr  = np.arange(SIZE)

t0 = time.perf_counter()
py_result = [x * 2 for x in py_list]
list_time = time.perf_counter() - t0

t0 = time.perf_counter()
np_result = np_arr * 2
np_time = time.perf_counter() - t0

print(f'Python list time : {list_time:.4f}s')
print(f'NumPy array time : {np_time:.4f}s')
print(f'NumPy is {list_time / np_time:.1f}x faster')
```

```python
# Memory comparison
print(f'Python list memory : {sys.getsizeof(py_list):,} bytes')
print(f'NumPy array memory : {np_arr.nbytes:,} bytes')
```
